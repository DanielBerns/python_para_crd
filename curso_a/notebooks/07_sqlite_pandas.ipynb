# Sesión 7: SQLite y Pandas

```python
import sqlite3
import pandas as pd
```

## Conectar a la base de datos

```python
# Creamos la conexión al archivo .sqlite
conn = sqlite3.connect('datos/estaciones_meteo.sqlite')
```

## Consultas SQL Directas (sqlite3)

```python
cursor = conn.cursor()
cursor.execute('SELECT * FROM info_estaciones')
estaciones = cursor.fetchall()
for est in estaciones:
    print(est)
```

## Integración con Pandas (Recomendado)

```python
# Leer tabla de mediciones directamente a un DataFrame
df_sql = pd.read_sql_query('SELECT * FROM mediciones LIMIT 10', conn)
display(df_sql)
```

```python
# Hacer un JOIN complejo usando SQL y traerlo a Pandas
query = '''
SELECT m.Fecha, e.nombre, m.Temperatura_Max, m.Precipitacion_mm
FROM mediciones m
JOIN info_estaciones e ON m.id_estacion = e.id_estacion
WHERE m.Temperatura_Max > 30
'''
df_dias_calurosos = pd.read_sql_query(query, conn)
display(df_dias_calurosos.head())
```

```python
# Cerrar conexión
conn.close()
```
