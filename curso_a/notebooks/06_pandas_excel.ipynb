# Sesión 6: Archivos Excel y Limpieza de Datos

```python
import pandas as pd
import numpy as np
```

## Leer Excel

```python
# Leemos el registro completo
df_meteo = pd.read_excel('datos/registro_meteorologico.xlsx')
display(df_meteo.head())
```

## Limpieza de Datos
En meteorología es muy común tener datos erróneos por falla de sensores.

```python
# Detectar NaN en temperatura máxima
faltantes = df_meteo['Temperatura_Max'].isna().sum()
print(f'Faltantes en Temp Max: {faltantes}')
```

```python
# Detectar humedad errónea (ej: -999)
errores_humedad = len(df_meteo[df_meteo['Humedad_Relativa'] < 0])
print(f'Errores en Humedad (-999): {errores_humedad}')
```

```python
# Limpiar: reemplazar -999 con NaN y luego imputar o borrar
df_limpio = df_meteo.copy()
df_limpio.loc[df_limpio['Humedad_Relativa'] < 0, 'Humedad_Relativa'] = np.nan
# Borramos las filas que contengan NaN para simplificar
df_limpio = df_limpio.dropna()
print('Datos originales:', len(df_meteo), 'Datos limpios:', len(df_limpio))
```

## Agrupación (Groupby)

```python
# Promedio de variables por estación
promedios = df_limpio.groupby('Estacion')[['Temperatura_Max', 'Temperatura_Min', 'Humedad_Relativa']].mean()
display(promedios)
```
