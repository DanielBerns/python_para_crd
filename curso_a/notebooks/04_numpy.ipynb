# Sesión 4: Introducción a NumPy
NumPy (Numerical Python) es la librería base para cálculos científicos en Python.

```python
import numpy as np
```

## Arreglos (Arrays)

```python
# Crear un arreglo a partir de una lista
temps_lista = [15.5, 16.2, 17.0, 14.8]
temps_array = np.array(temps_lista)
print(temps_array)
print('Tipo:', type(temps_array))
```

```python
# Operaciones vectorizadas (mucho más rápido que un for)
temps_kelvin = temps_array + 273.15
print('En Kelvin:', temps_kelvin)
```

## Estadísticas básicas

```python
print('Media:', np.mean(temps_array))
print('Máxima:', np.max(temps_array))
print('Desvío estándar:', np.std(temps_array))
```

```python
# Manejo de valores faltantes (NaN)
temps_con_nan = np.array([15.5, np.nan, 17.0])
print('Media ignorando NaN:', np.nanmean(temps_con_nan))
```
