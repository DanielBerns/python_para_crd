### Desafío 1.1: Conversor de Fahrenheit a Celsius
La fórmula para convertir grados Fahrenheit ($F$) a Celsius ($C$) es:
$$C = \frac{F - 32}{1.8}$$
Completa la fórmula matemática en la variable `temperatura_celsius` para convertir `temperatura_fahrenheit` (que vale 77).

```python
temperatura_fahrenheit = 77.0
# REEMPLAZA EL 0.0 CON TU FÓRMULA MATEMÁTICA
temperatura_celsius = (temperatura_fahrenheit - 32) / 1.8
print(f'{temperatura_fahrenheit}°F equivale a {temperatura_celsius}°C')
```

```python
# CELDA DE VERIFICACIÓN - ¡No la modifiques!
try:
    assert abs(temperatura_celsius - 25.0) < 1e-5, 'La conversión no es correcta. Revisa la fórmula.'
    print('¡Felicidades! Desafío 1.1 superado con éxito. 🎉')
except AssertionError as e:
    print(f'Error: {e}')
```
