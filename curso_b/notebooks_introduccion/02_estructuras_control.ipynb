## 1. Listas y Diccionarios
* Las **listas** son secuencias ordenadas de elementos que se definen con corchetes `[]`.
* Los **diccionarios** guardan datos en pares clave-valor utilizando llaves `{}`.

```python
temperaturas = [12.5, 13.0, 14.2, 11.8]
datos_estacion = {
    'nombre': 'Ezeiza',
    'latitud': -34.8,
    'longitud': -58.5
}
print(f'Primera temperatura: {temperaturas[0]}')
print(f'Estación: {datos_estacion.get("nombre")}')
```

## 2. Desafío Práctico: Filtrado de Heladas
### Desafío 2.1: Filtrando Heladas
Completa el código a continuación usando un bucle `for` y una estructura condicional `if` para encontrar la cantidad de temperaturas bajo cero (heladas).

```python
registro_semanal = [2.5, -1.2, 0.4, -3.1, -0.8, 4.2, 5.0]
cantidad_heladas = 0

# TU CÓDIGO AQUÍ: Recorre la lista 'registro_semanal' e incrementa 'cantidad_heladas' si el valor es menor que 0
for temp in registro_semanal:
    if temp < 0:
        cantidad_heladas += 1

print(f'Días con heladas registrados: {cantidad_heladas}')
```

```python
# CELDA DE VERIFICACIÓN
try:
    assert cantidad_heladas == 3, f'La respuesta es incorrecta. Se esperaban 3 heladas, obtuviste {cantidad_heladas}.'
    print('¡Felicidades! Desafío 2.1 superado con éxito. ❄️')
except AssertionError as e:
    print(f'Error: {e}')
```
