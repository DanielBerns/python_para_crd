# Sesión 8: Visualización de Datos con Matplotlib y Seaborn

La visualización es una de las herramientas más potentes en ciencia de datos para identificar patrones, tendencias y anomalías en series de tiempo climáticas. En esta sesión aprenderás a:
* Crear gráficos de líneas para representar series de tiempo.
* Diseñar gráficos de dispersión (*scatter plots*) para analizar correlaciones.
* Personalizar elementos del gráfico (títulos, etiquetas, leyendas y cuadrícula).

## 1. Conceptos Básicos de Graficación
La biblioteca **Matplotlib** (específicamente el módulo `pyplot`) provee una interfaz para construir gráficos paso a paso, mientras que **Seaborn** simplifica la creación de gráficos estadísticos atractivos sobre estructuras de Pandas.

```python
import matplotlib.pyplot as plt
import pandas as pd

# Datos de ejemplo para graficar
dias = ['Lun', 'Mar', 'Mie', 'Jue', 'Vie']
temp_promedio = [18.2, 19.5, 17.0, 16.5, 21.0]

plt.figure(figsize=(8, 4))
plt.plot(dias, temp_promedio, marker='o', color='tab:red', linestyle='--')
plt.title('Evolución de la Temperatura Semanal')
plt.xlabel('Días de la semana')
plt.ylabel('Temperatura (°C)')
plt.grid(True, linestyle=':')
plt.show()
```

### Desafío 8.1: Gráfico de Líneas de Temperatura Mínima vs Máxima
Completa el código a continuación para generar un gráfico de líneas comparativo. Debe graficar la evolución de la `'temperatura_min'` (color azul) y la `'temperatura_max'` (color rojo) a lo largo de los índices del DataFrame `df_clima`.

```python
import pandas as pd
import matplotlib.pyplot as plt

df_clima = pd.read_csv('/content/notebooks_introduccion/datos_basicos.csv')

fig, ax = plt.subplots(figsize=(10, 5))

# TU CÓDIGO AQUÍ:
# 1. Grafica 'temperatura_min' con color azul ('blue') y etiqueta 'Temp Mínima'
ax.plot(df_clima.index, df_clima['temperatura_min'], color='blue', label='Temp Mínima')

# 2. Grafica 'temperatura_max' con color rojo ('red') y etiqueta 'Temp Máxima'
ax.plot(df_clima.index, df_clima['temperatura_max'], color='red', label='Temp Máxima')

# Personalización del gráfico
ax.set_title('Historial de Temperaturas Mínimas y Máximas')
ax.set_xlabel('Índice del Registro')
ax.set_ylabel('Temperatura (°C)')
ax.legend()
ax.grid(True)
plt.show()
```
