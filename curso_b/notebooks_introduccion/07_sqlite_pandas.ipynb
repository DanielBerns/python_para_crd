# Sesión 7: Integración de SQL con Pandas

Las bases de datos relacionales son el estándar industrial para almacenar grandes volúmenes de datos estructurados. En esta sesión aprenderás a conectar Python con una base de datos local **SQLite** y extraer información directamente a DataFrames de **Pandas** usando consultas SQL SQL de selección (`SELECT`).

## 1. Conexión a SQLite y Carga con Pandas
El módulo integrado `sqlite3` permite crear una conexión con la base de datos. Una vez establecida la conexión, Pandas nos facilita la vida con la función `pd.read_sql_query()`, la cual ejecuta la consulta y nos devuelve un DataFrame estructurado de forma inmediata.

```python
import sqlite3
import pandas as pd

# Conectarse a la base de datos
conexion = sqlite3.connect('/content/notebooks_introduccion/estaciones_meteo.sqlite')

# Ver las tablas disponibles
tablas = pd.read_sql_query("SELECT name FROM sqlite_master WHERE type='table';", conexion)
print('Tablas en la base de datos:')
display(tablas)
conexion.close()
```

## 2. Desafío Práctico: Consulta y Extracción de Datos
Trabajaremos con la tabla `estaciones` que se encuentra dentro de nuestra base de datos relacional.

### Desafío 7.1: Consulta SQL de Estaciones por Provincia
Escribe y ejecuta una consulta SQL para seleccionar todos los campos de la tabla `estaciones` donde la provincia (columna `'provincia'`) sea igual a `'Buenos Aires'`. Carga el resultado en un DataFrame llamado `df_buenos_aires`.

```python
import sqlite3
import pandas as pd

# 1. Establecer conexión
conn = sqlite3.connect('/content/notebooks_introduccion/estaciones_meteo.sqlite')

# TU CÓDIGO AQUÍ:
# 2. Define la consulta SQL
consulta_sql = "SELECT * FROM estaciones WHERE provincia = 'Buenos Aires'"

# 3. Cargar a DataFrame
df_buenos_aires = pd.read_sql_query(consulta_sql, conn)

# Cerrar la conexión
conn.close()

print(f'Cantidad de estaciones en Buenos Aires encontradas: {len(df_buenos_aires)}')
display(df_buenos_aires.head())
```

```python
# CELDA DE VERIFICACIÓN
try:
    assert 'df_buenos_aires' in globals(), 'El DataFrame df_buenos_aires no está definido.'
    assert isinstance(df_buenos_aires, pd.DataFrame), 'df_buenos_aires debe ser un DataFrame de Pandas.'
    assert not df_buenos_aires.empty, 'El DataFrame está vacío. Verifica tu consulta SQL.'
    assert (df_buenos_aires['provincia'] == 'Buenos Aires').all(), 'La consulta extrajo estaciones de otras provincias.'
    print('¡Extraordinario! Desafío 7.1 (SQL + Pandas) superado y verificado con éxito. 🗄️')
except AssertionError as e:
    print(f'Error de validación: {e}')
```
