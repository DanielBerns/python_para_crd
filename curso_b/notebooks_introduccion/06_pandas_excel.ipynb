### Desafío 6.1: Lectura, Cálculo y Exportación
Completa los siguientes pasos:
1. Carga el archivo `'registro_meteorologico.xlsx'` en un DataFrame llamado `df_excel`.
2. Calcula la temperatura promedio (columna `'Temperatura'`) para cada `'Estacion'` utilizando el agrupamiento `.groupby('Estacion')['Temperatura'].mean()`. Guarda este resultado (que será una Serie) en la variable `promedio_por_estacion`.
3. Exporta la serie resultante `promedio_por_estacion` a un nuevo archivo de Excel llamado `'resumen_temperaturas.xlsx'`.

```python
import pandas as pd
import os

ruta_excel = '/content/notebooks_introduccion/registro_meteorologico.xlsx'

# TU CÓDIGO AQUÍ:
# 1. Cargar el DataFrame
df_excel = pd.read_excel(ruta_excel)

# 2. Agrupar y promediar la temperatura
promedio_por_estacion = df_excel.groupby('Estacion')['Temperatura'].mean()

# 3. Exportar el resultado a Excel
promedio_por_estacion.to_excel('resumen_temperaturas.xlsx')

print('Resumen de promedios por estación:')
print(promedio_por_estacion)
```

```python
# CELDA DE VERIFICACIÓN
try:
    assert 'df_excel' in globals(), 'No se cargó el DataFrame df_excel.'
    assert 'promedio_por_estacion' in globals(), 'La variable promedio_por_estacion no está definida.'
    assert os.path.exists('resumen_temperaturas.xlsx'), 'No se generó el archivo de salida resumen_temperaturas.xlsx.'
    # Comprobar algunos valores promedio característicos
    assert abs(promedio_por_estacion['Aeroparque'] - 18.0) < 1.0, 'El promedio calculado para Aeroparque no es correcto.'
    print('¡Espectacular! Desafío 6.1 (Integración con Excel) superado y verificado. 📊')
except AssertionError as e:
    print(f'Error de validación: {e}')
```
