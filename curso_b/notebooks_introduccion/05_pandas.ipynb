### Desafío 5.1: Análisis Exploratorio del Dataset
Utilizando el DataFrame `df_clima` cargado anteriormente, completa el código para calcular:
1. La temperatura promedio máxima (columna `'temperatura_max'`) y guárdala en la variable `temp_max_promedio`.
2. Filtra el DataFrame para obtener únicamente los registros donde la precipitación (columna `'precipitacion'`) sea estrictamente mayor a 10 mm. Guarda este DataFrame filtrado en la variable `df_lluvia_intensa`.

```python
import pandas as pd
ruta_csv = '/content/notebooks_introduccion/datos_basicos.csv'
df_clima = pd.read_csv(ruta_csv)

# TU CÓDIGO AQUÍ:
# 1. Calcula el promedio de la columna 'temperatura_max'
temp_max_promedio = df_clima['temperatura_max'].mean()

# 2. Filtra las filas donde 'precipitacion' > 10.0
df_lluvia_intensa = df_clima[df_clima['precipitacion'] > 10.0]

print(f'Temperatura máxima promedio: {temp_max_promedio:.2f}°C')
print(f'Cantidad de registros con lluvia intensa: {len(df_lluvia_intensa)}')
```
