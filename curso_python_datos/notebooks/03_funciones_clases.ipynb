## Funciones
Las funciones agrupan código que podemos llamar múltiples veces.

```python
def celsius_a_fahrenheit(c):
    return (c * 9/5) + 32

print('20°C en F:', celsius_a_fahrenheit(20))
```

## Clases y Objetos
Una clase es un 'molde' para crear objetos. Librerías como Pandas usan clases constantemente.

```python
class EstacionMeteorologica:
    def __init__(self, nombre, lat, lon):
        self.nombre = nombre
        self.lat = lat
        self.lon = lon
        self.mediciones = []

    def agregar_medicion(self, temp):
        self.mediciones.append(temp)

    def promedio(self):
        return sum(self.mediciones) / len(self.mediciones) if self.mediciones else 0
```

```python
est_eze = EstacionMeteorologica('Ezeiza', -34.82, -58.53)
est_eze.agregar_medicion(20)
est_eze.agregar_medicion(22)
print(f'Promedio de {est_eze.nombre}: {est_eze.promedio()}')
```
