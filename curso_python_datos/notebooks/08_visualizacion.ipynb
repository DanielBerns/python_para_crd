# Sesión 8: Visualización de Datos
En meteorología es fundamental graficar las series temporales.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Configuración para ver gráficos dentro del notebook
%matplotlib inline
```

## Cargar Datos Limpios

```python
# Volvemos a leer y limpiar rápidamente
df_meteo = pd.read_excel('datos/registro_meteorologico.xlsx')
df_meteo = df_meteo[df_meteo['Humedad_Relativa'] >= 0]
df_meteo = df_meteo.dropna()
df_meteo['Fecha'] = pd.to_datetime(df_meteo['Fecha'])
```

## Filtrar para una estación y fijar índice

```python
df_aero = df_meteo[df_meteo['Estacion'] == 'Estacion_A_Aeroparque'].copy()
df_aero.set_index('Fecha', inplace=True)
display(df_aero.head())
```

## Gráficos usando Pandas/Matplotlib

```python
# Plot simple de Temperatura Máxima (Serie temporal)
plt.figure(figsize=(12, 5))
df_aero['Temperatura_Max'].plot(color='red', title='Temperatura Máxima en Aeroparque (2023)')
plt.ylabel('Temperatura (°C)')
plt.grid(True)
plt.show()
```

```python
# Histograma de humedad
plt.figure(figsize=(8, 5))
df_aero['Humedad_Relativa'].plot(kind='hist', bins=20, color='blue', alpha=0.7)
plt.title('Distribución de la Humedad Relativa')
plt.xlabel('Humedad (%)')
plt.show()
```

```python
# Scatter plot (Dispersión): T_Max vs Humedad
df_aero.plot.scatter(x='Temperatura_Max', y='Humedad_Relativa', figsize=(8,5), alpha=0.5)
plt.title('Relación entre T. Máxima y Humedad')
plt.show()
```
