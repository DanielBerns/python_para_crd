## Variables y Tipos de Datos
Podemos almacenar datos (como temperaturas, presiones, nombres de estaciones) en variables.

```python
temperatura = 23.5 # Float (número real)
estacion = 'Aeroparque' # String (cadena de texto)
lluvia = True # Booleano (Verdadero o Falso)
print(f'Estación {estacion}: Temperatura={temperatura}C, Lluvia={lluvia}')
```

## Operadores Matemáticos

```python
t_min = 15.2
t_max = 28.4
amplitud_termica = t_max - t_min
print('Amplitud Térmica:', amplitud_termica)
```
