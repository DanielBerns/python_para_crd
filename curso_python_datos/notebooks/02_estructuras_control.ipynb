# Sesión 2: Estructuras y Control de Flujo

```python
# Listas de temperaturas
temps = [12.5, 13.0, 14.2, 11.8]
temps.append(15.1)
print('Temperaturas:', temps)
```

```python
# Diccionarios (muy útiles para datos estructurados)
clima = {'estacion': 'Ezeiza', 'temp_max': 25, 'temp_min': 15}
print('Estación:', clima['estacion'])
```

## Condicionales (if / else)

```python
temp = 32
if temp > 30:
    print('Alerta de calor')
elif temp < 5:
    print('Alerta de helada')
else:
    print('Temperatura normal')
```

## Bucles (for)

```python
# Calcular el promedio de una lista
suma = 0
for t in temps:
    suma += t
promedio = suma / len(temps)
print('Temperatura promedio:', promedio)
```
